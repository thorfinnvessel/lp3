```python
# Step 1: Import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Step 2: Define the function
def func1(x):
    return (x + 3) ** 2
```

```python
# Step 3: Define the gradient (derivative)
def gradient_func1(x):
    return 2 * (x + 3)
```

```python
# Step 4: Define Gradient Descent
def gradient_descent(function, start, learn_rate, n_iter=100):

    gradient = gradient_func1

    points = [start]
    iters = 0
    
    # Step 5: Update x using gradient descent
    while iters < n_iter:
        prev_x = start
        start = start - learn_rate * gradient(prev_x)
        iters = iters + 1
        points.append(start)

    # Step 6: Print minimum
    print("The local minimum occurs at", start)

    # Step 7: Generate values for plotting
    x_values = np.linspace(-7, 5, 100)

    y_values = function(x_values)

    # Step 8: Plot the function
    plt.figure(figsize=(10, 6))

    plt.plot(x_values, y_values, 'g', label='Function')

    plt.plot(
        points,
        function(np.array(points)),
        '-o',
        label='Gradient Descent'
    )

    plt.xlabel("X")
    plt.ylabel("f(X)")
    plt.title("Gradient Descent Optimization")

    plt.legend()
    plt.grid()

    plt.show()

# Step 9: Call the function
gradient_descent(
    function=func1,
    start=2.0,
    learn_rate=0.2,
    n_iter=100
)
```
